```python
import pandas as pd
from pathlib import Path

# ディレクトリと出力ファイル
indir = Path(r"out\new_attack")
outfile = indir / "F22.csv"

# Fij_01.csv ～ Fij_20.csv のパス
files = [indir / f"Fij_new_{i:02d}.csv" for i in range(1, 21)]

# 各ファイルを読み込んで列方向に結合
dfs = [pd.read_csv(f, header=None) for f in files]
merged = pd.concat(dfs, axis=1)

# 保存（ヘッダもインデックスも出力しない）
merged.to_csv(outfile, index=False, header=False)

print(f"Saved {outfile}, shape={merged.shape}")
```
